## 3. Create
Criação de múltiplos registros para cada entidade do domínio.

```java
List<Cliente> clientes = new ArrayList<>();
String[] nomes = {"João Silva", "Maria Souza", "Carlos Pereira"};
String[] cpfs = {"111.111.111-11", "222.222.222-22", "333.333.333-33"};

for (int i = 0; i < nomes.length; i++) {
    Cliente c = new Cliente();
    c.setNome(nomes[i]);
    c.setCpfCnpj(cpfs[i]);
    c.setEndereco("Rua Exemplo");
    c.setCep("74000-00" + i);
    c.setNumero(String.valueOf(i + 1));
    c.setComplemento("Apto " + (i + 1));
    clienteRepo.create(c);
    clientes.add(c);
    System.out.println("Cliente criado: " + c.getNome() + " id=" + c.getId());
}
```

```java
List<Veiculo> veiculos = new ArrayList<>();
String[] placas = {"ABC-1234", "DEF-5678", "GHI-9012"};
String[] modelos = {"Gol", "Civic", "Corolla"};
int[] anos = {2020, 2021, 2022};

for (int i = 0; i < placas.length; i++) {
    Veiculo v = new Veiculo();
    v.setPlaca(placas[i]);
    v.setModelo(modelos[i]);
    v.setAnoModelo(anos[i]);
    v.setCliente(clientes.get(i));
    veiculoRepo.create(v);
    veiculos.add(v);
    System.out.println("Veiculo criado: " + v.getModelo() + " id=" + v.getId());
    }
```

```java
List<Mecanico> mecanicos = new ArrayList<>();
String[] nomesM = {"Carlos Souza", "Pedro Lima", "Ana Paula"};

for (String nome : nomesM) {
    Mecanico m = new Mecanico();
    m.setNome(nome);
    mecanicoRepo.create(m);
    mecanicos.add(m);
    System.out.println("Mecanico criado: " + m.getNome() + " id=" + m.getId());
}
```

```java
List<Servico> servicos = new ArrayList<>();
String[] descricoes = {"Troca de oleo", "Alinhamento", "Balanceamento"};
double[] valores = {80.0, 60.0, 50.0};

for (int i = 0; i < descricoes.length; i++) {
    Servico s = new Servico();
    s.setDescricao(descricoes[i]);
    s.setValor(valores[i]);
    servicoRepo.create(s);
    servicos.add(s);
    System.out.println("Servico criado: " + s.getDescricao() + " id=" + s.getId());
}
```

```java
List<Peca> pecas = new ArrayList<>();
String[] nomesPeca = {"Filtro de oleo", "Vela de ignicao", "Pastilha de freio"};
double[] precos = {35.0, 20.0, 90.0};

for (int i = 0; i < nomesPeca.length; i++) {
    Peca p = new Peca();
    p.setNome(nomesPeca[i]);
    p.setPrecoUnitario(precos[i]);
    pecaRepo.create(p);
    pecas.add(p);
    System.out.println("Peca criada: " + p.getNome() + " id=" + p.getId());
}
```

```java
List<OrdemServico> ordens = new ArrayList<>();
double[] valores2 = {150.0, 200.0, 250.0};

for (int i = 0; i < veiculos.size(); i++) {
    OrdemServico o = new OrdemServico();
    o.setDataEmissao(new Date());
    o.setValorTotal(valores2[i]);
    o.setVeiculo(veiculos.get(i));
    ordemRepo.create(o);
    ordens.add(o);
    System.out.println("Ordem criada id=" + o.getId() + " veiculo=" + o.getVeiculo().getModelo());
}
```

```java
for (int i = 0; i < ordens.size(); i++) {
    OrdemMecanico om1 = new OrdemMecanico();
    om1.setOrdemServico(ordens.get(i));
    om1.setMecanico(mecanicos.get(i % mecanicos.size()));
    ordemMecanicoRepo.create(om1);

    OrdemMecanico om2 = new OrdemMecanico();
    om2.setOrdemServico(ordens.get(i));
    om2.setMecanico(mecanicos.get((i + 1) % mecanicos.size()));
    ordemMecanicoRepo.create(om2);

    System.out.println("Ordem " + ordens.get(i).getId() + " vinculada a 2 mecanicos");
}
```

```java
for (int i = 0; i < ordens.size(); i++) {
    ItemServico is = new ItemServico();
    is.setQuantidade(i + 1);
    is.setServico(servicos.get(i % servicos.size()));
    is.setOrdemServico(ordens.get(i));
    itemServicoRepo.create(is);
    System.out.println("ItemServico criado: ordem=" + ordens.get(i).getId() + " servico=" + servicos.get(i % servicos.size()).getDescricao());
}
```

```java
for (int i = 0; i < ordens.size(); i++) {
    ItemPeca ip = new ItemPeca();
    ip.setQuantidade(i + 1);
    ip.setPeca(pecas.get(i % pecas.size()));
    ip.setOrdemServico(ordens.get(i));
    itemPecaRepo.create(ip);
    System.out.println("ItemPeca criado: ordem=" + ordens.get(i).getId() + " peca=" + pecas.get(i % pecas.size()).getNome());
}
```

```java
for (int i = 0; i < ordens.size(); i++) {
    NotaFiscal nf = new NotaFiscal();
    nf.setNumero("NF-00" + (i + 1));
    nf.setChaveAcesso("CHAVE" + String.format("%039d", i + 1));
    nf.setDataEmissao(new Date());
    nf.setValorImposto(ordens.get(i).getValorTotal() * 0.1);
    nf.setOrdemServico(ordens.get(i));
    notaRepo.create(nf);
    System.out.println("Nota criada: " + nf.getNumero() + " imposto=R$" + nf.getValorImposto());
}
```

## 5. Update
Atualização de campos em todas as entidades.

```java
clienteRepo.findAll().forEach(c -> {
    c.setNome(c.getNome() + " [atualizado]");
    clienteRepo.update(c);
});

veiculoRepo.findAll().forEach(v -> {
    v.setAnoModelo(v.getAnoModelo() + 1);
    veiculoRepo.update(v);
});

ordemRepo.findAll().forEach(o -> {
    o.setValorTotal(o.getValorTotal() + 50.0);
    ordemRepo.update(o);
});

mecanicoRepo.findAll().forEach(m -> {
    m.setNome(m.getNome() + " [atualizado]");
    mecanicoRepo.update(m);
});

servicoRepo.findAll().forEach(s -> {
    s.setValor(s.getValor() + 10.0);
    servicoRepo.update(s);
});

pecaRepo.findAll().forEach(p -> {
    p.setPrecoUnitario(p.getPrecoUnitario() + 5.0);
    pecaRepo.update(p);
});

itemServicoRepo.findAll().forEach(is -> {
    is.setQuantidade(is.getQuantidade() + 1);
    itemServicoRepo.update(is);
});

itemPecaRepo.findAll().forEach(ip -> {
    ip.setQuantidade(ip.getQuantidade() + 1);
    itemPecaRepo.update(ip);
});

notaRepo.findAll().forEach(n -> {
    n.setValorImposto(n.getValorImposto() + 5.0);
    notaRepo.update(n);
});

ordemMecanicoRepo.findAll().forEach(om -> {
    try {
        List<Mecanico> todos = mecanicoRepo.findAll();
        om.setMecanico(todos.get((int)(om.getId() % todos.size())));
        ordemMecanicoRepo.update(om);
    } catch (Exception e) {
        System.out.println(e);
    }
});

System.out.println("Todas as entidades atualizadas!");
```
