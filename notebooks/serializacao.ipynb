# Oficina Mecânica - Serialização JSON
Exportação dos dados persistidos no banco SQLite para o formato JSON utilizando a biblioteca Gson.

## 1. Configuração
Carregamento das dependências Maven e importação das classes do projeto.

```java
%maven com.j256.ormlite:ormlite-core:6.1
%maven com.j256.ormlite:ormlite-jdbc:6.1
%maven org.xerial:sqlite-jdbc:3.45.1.0
%maven org.slf4j:slf4j-api:2.0.12
%maven ch.qos.logback:logback-classic:1.4.14
%maven com.google.code.gson:gson:2.11.0
```

```java
%classpath add jar ../target/classes

import br.edu.ufg.oficina.domain.*;
import br.edu.ufg.oficina.repository.*;
import com.google.gson.Gson;
import com.google.gson.GsonBuilder;
import java.io.FileWriter;
import java.io.FileReader;
import java.util.List;

System.out.println("Classes importadas com sucesso!");
```

## 2. Conexão com o Banco
Abertura da conexão com o banco e inicialização dos repositórios.

```java
Database database = new Database("oficina.db");
database.getConnection();

ClienteRepository clienteRepo = new ClienteRepository(database);
VeiculoRepository veiculoRepo = new VeiculoRepository(database);
MecanicoRepository mecanicoRepo = new MecanicoRepository(database);
ServicoRepository servicoRepo = new ServicoRepository(database);
PecaRepository pecaRepo = new PecaRepository(database);
OrdemServicoRepository ordemRepo = new OrdemServicoRepository(database);
OrdemMecanicoRepository ordemMecanicoRepo = new OrdemMecanicoRepository(database);
ItemServicoRepository itemServicoRepo = new ItemServicoRepository(database);
ItemPecaRepository itemPecaRepo = new ItemPecaRepository(database);
NotaFiscalRepository notaRepo = new NotaFiscalRepository(database);

System.out.println("Repositorios inicializados!");
```

## 3. Serialização
Conversão dos objetos persistidos no banco para o formato JSON.

```java
Gson gson = new GsonBuilder()
        .setDateFormat("yyyy-MM-dd")
    .setPrettyPrinting()
    .create();

System.out.println("Gson configurado!");
```

```java
List<Cliente> clientes = clienteRepo.findAll();
List<Veiculo> veiculos = veiculoRepo.findAll();
List<Mecanico> mecanicos = mecanicoRepo.findAll();
List<Servico> servicos = servicoRepo.findAll();
List<Peca> pecas = pecaRepo.findAll();
List<OrdemServico> ordens = ordemRepo.findAll();
List<OrdemMecanico> ordemMecanicos = ordemMecanicoRepo.findAll();
List<ItemServico> itemServicos = itemServicoRepo.findAll();
List<ItemPeca> itemPecas = itemPecaRepo.findAll();
List<NotaFiscal> notas = notaRepo.findAll();

System.out.println("=== Clientes ===");
System.out.println(gson.toJson(clientes));

System.out.println("=== Veiculos ===");
System.out.println(gson.toJson(veiculos));

System.out.println("=== Mecanicos ===");
System.out.println(gson.toJson(mecanicos));

System.out.println("=== Servicos ===");
System.out.println(gson.toJson(servicos));

System.out.println("=== Pecas ===");
System.out.println(gson.toJson(pecas));

System.out.println("=== Ordens de Servico ===");
System.out.println(gson.toJson(ordens));

System.out.println("=== OrdemMecanico ===");
System.out.println(gson.toJson(ordemMecanicos));

System.out.println("=== Itens de Servico ===");
System.out.println(gson.toJson(itemServicos));

System.out.println("=== Itens de Peca ===");
System.out.println(gson.toJson(itemPecas));

System.out.println("=== Notas Fiscais ===");
System.out.println(gson.toJson(notas));
```

## 4. Exportação para Arquivo
Salvamento dos dados serializados em arquivos JSON.

```java
String[] nomes = {"clientes", "veiculos", "mecanicos", "servicos", "pecas", "ordens", "ordem_mecanico", "item_servico", "item_peca", "notas"};
List[] listas = {clientes, veiculos, mecanicos, servicos, pecas, ordens, ordemMecanicos, itemServicos, itemPecas, notas};

for (int i = 0; i < nomes.length; i++) {
    String caminho = "../notebooks/" + nomes[i] + ".json";
    FileWriter writer = new FileWriter(caminho);
    writer.write(gson.toJson(listas[i]));
    writer.close();
    System.out.println("Salvo: " + caminho);
}
```

## 5. Desserialização
Leitura dos arquivos JSON e reconstrução dos objetos Java. 

**Obs.:** A desserialização realizada neste notebook reconstrói os objetos Java em memória a partir dos arquivos JSON, mas **não os persiste novamente no banco de dados**.
O motivo é que os objetos possuem relacionamentos entre si (ex: `OrdemMecanico` referencia `OrdemServico` e `Mecanico`). Ao repersistir, o ORMLite geraria novos IDs para cada objeto, quebrando essas referências e corrompendo a integridade dos dados.
O objetivo deste notebook é demonstrar que os dados podem ser **serializados e desserializados corretamente**, garantindo que nenhuma informação é perdida no processo.

```java
FileReader reader = new FileReader("../notebooks/clientes.json");
Cliente[] clientesLidos = gson.fromJson(reader, Cliente[].class);
reader.close();

System.out.println("Clientes desserializados:");
for (Cliente c : clientesLidos) {
    System.out.println(c.getId() + " - " + c.getNome());
}
```

```java
FileReader r2 = new FileReader("../notebooks/veiculos.json");
Veiculo[] veiculosLidos = gson.fromJson(r2, Veiculo[].class);
r2.close();

FileReader r3 = new FileReader("../notebooks/mecanicos.json");
Mecanico[] mecanicosLidos = gson.fromJson(r3, Mecanico[].class);
r3.close();

FileReader r4 = new FileReader("../notebooks/servicos.json");
Servico[] servicosLidos = gson.fromJson(r4, Servico[].class);
r4.close();

FileReader r5 = new FileReader("../notebooks/pecas.json");
Peca[] pecasLidas = gson.fromJson(r5, Peca[].class);
r5.close();

FileReader r6 = new FileReader("../notebooks/ordens.json");
OrdemServico[] ordensLidas = gson.fromJson(r6, OrdemServico[].class);
r6.close();

FileReader r7 = new FileReader("../notebooks/ordem_mecanico.json");
OrdemMecanico[] ordemMecanicosLidos = gson.fromJson(r7, OrdemMecanico[].class);
r7.close();

FileReader r8 = new FileReader("../notebooks/item_servico.json");
ItemServico[] itemServicosLidos = gson.fromJson(r8, ItemServico[].class);
r8.close();

FileReader r9 = new FileReader("../notebooks/item_peca.json");
ItemPeca[] itemPecasLidas = gson.fromJson(r9, ItemPeca[].class);
r9.close();

FileReader r10 = new FileReader("../notebooks/notas.json");
NotaFiscal[] notasLidas = gson.fromJson(r10, NotaFiscal[].class);
r10.close();

System.out.println("Veiculos desserializados: " + veiculosLidos.length);
System.out.println("Mecanicos desserializados: " + mecanicosLidos.length);
System.out.println("Servicos desserializados: " + servicosLidos.length);
System.out.println("Pecas desserializadas: " + pecasLidas.length);
System.out.println("Ordens desserializadas: " + ordensLidas.length);
System.out.println("OrdemMecanico desserializados: " + ordemMecanicosLidos.length);
System.out.println("ItemServicos desserializados: " + itemServicosLidos.length);
System.out.println("ItemPecas desserializadas: " + itemPecasLidas.length);
System.out.println("Notas desserializadas: " + notasLidas.length);
```

Fechamento da conexão com o banco.
